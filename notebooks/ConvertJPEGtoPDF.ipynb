# Convert Images to PDF


## How to use


To **convert** a set of JPG/JPEG images into a single PDF file:  
1) Prepare the image files (.jpg or .jpeg) that you want to convert. They will be combined into the PDF in alphabetical order by file name. Rename files if needed to control the order.  
2) Upload all image files that you want to convert into the root directory of the Files area.  
3) In the **Set parameters** section, set the preferred name for the resulting pdf file.  
4) In the **Set parameters** section, choose the page dimensions mode:  
   * *fixed* - every page uses the same fixed size. Then choose *letter_portrait* or *letter_landscape*.  
   * *custom* - every page uses the same custom size. Then set the height and width, in inches.  
   * *derived_from_image* - each page is sized to match its own image, using the image's embedded DPI (dots per inch) if it has one, or a default of 96 DPI if it does not.  
5) In the **Set parameters** section, choose whether to optimize the resulting file size:  
   * *not_optimize* - images are embedded without recompression, so the output file size stays close to the combined size of the source images.  
   * *optimize* - every page is re-rendered at a target DPI and re-compressed as a JPEG at a target quality, which can shrink the file substantially. Then set:  
     * the target DPI - a preset (300, 150, or 96) or a custom value. Setting a target DPI lower than the effective DPI of your source images is what actually reduces file size; a target DPI at or above it will not shrink the file (it may even grow slightly) since it only adds a decode/recompress step.  
     * the target JPEG quality (1-100, default 75). Lower values shrink the file further at the cost of image quality.  
6) Run all cells in the notebook (Runtime - Run all or Ctrl-F9).  
7) Download the output pdf from the Files area (Refresh to see the newly created file).

If a file with the chosen output name already exists, no action is taken. Delete or rename it first.

```python
import os

import img2pdf
import pymupdf
```

### Collect image files


```python
def get_image_files():
    """
    Get all jpg/jpeg files in the current directory, sorted alphabetically.

    Return a tuple (list of file names, message string).
    If no matching files are found, the list is empty and the message
    explains that no action was taken.
    """
    image_files = []
    for filename in os.listdir():
        if filename.lower().endswith(('.jpg', '.jpeg')):
            image_files.append(filename)

    if len(image_files) == 0:
        return [], "No jpg/jpeg files found. No action taken."

    # sort image_files in the alphabetical order
    image_files.sort()

    return image_files, "OK"
```

### Determine the page layout


```python
def get_layout_fun():
    """
    Build the img2pdf layout function according to page_size_mode.

    Return a layout function to pass to img2pdf.convert(), or None to
    use img2pdf's default layout. The default layout sizes each page
    to match its own image: it uses the image's embedded DPI (dots per
    inch) if the image has one, otherwise it assumes 96 DPI.
    """
    if page_size_mode == "derived_from_image":
        return None

    if page_size_mode == "fixed":
        if fixed_page_size == "letter_portrait":
            page_width_pt = img2pdf.in_to_pt(8.5)
            page_height_pt = img2pdf.in_to_pt(11)
        else:
            page_width_pt = img2pdf.in_to_pt(11)
            page_height_pt = img2pdf.in_to_pt(8.5)
    else:
        # page_size_mode == "custom"
        page_width_pt = img2pdf.in_to_pt(custom_width_in)
        page_height_pt = img2pdf.in_to_pt(custom_height_in)

    # fit each image within the page, preserving its aspect ratio, centered
    return img2pdf.get_layout_fun(
        pagesize=(page_width_pt, page_height_pt),
        fit=img2pdf.FitMode.into,
    )
```

### Build the PDF


```python
def build_pdf_bytes(image_files):
    """
    Convert a list of image files into PDF bytes, one image per page,
    according to page_size_mode. Images are embedded without
    recompression at this stage.

    :param image_files: List of image file names, in the order they
    should appear in the PDF.
    Return the PDF content as bytes.
    """
    layout_fun = get_layout_fun()

    # only pass layout_fun when it is set, so that img2pdf's own default
    # layout (used for the derived_from_image mode) applies otherwise
    convert_kwargs = {}
    if layout_fun is not None:
        convert_kwargs['layout_fun'] = layout_fun

    return img2pdf.convert(image_files, **convert_kwargs)
```

```python
def get_target_dpi():
    """
    Get the target DPI to use when optimizing, from the Set parameters
    section (preset_target_dpi or custom_target_dpi).

    Return the DPI as a float, regardless of whether the underlying
    parameter came through as a number or as a numeric string (a form
    field can sometimes yield a string after the dropdown is edited).
    """
    if target_dpi_mode == "preset":
        raw_dpi = preset_target_dpi
    else:
        raw_dpi = custom_target_dpi

    return float(raw_dpi)
```

```python
def downsample_pdf_bytes(pdf_bytes, target_dpi, jpeg_quality):
    """
    Shrink a PDF by re-rendering every page at target_dpi and
    re-compressing it as a JPEG at jpeg_quality. Each page keeps its
    original width and height; only the embedded image resolution and
    compression change.

    :param pdf_bytes: The source PDF, as bytes.
    :param target_dpi: The DPI to render each page at before recompressing.
    :param jpeg_quality: The JPEG quality (1-100) to recompress with.
    Return the optimized PDF as bytes.
    """
    src = pymupdf.open(stream=pdf_bytes, filetype="pdf")
    out = pymupdf.open()

    zoom = float(target_dpi) / 72.0  # PDF base is 72 dpi
    matrix = pymupdf.Matrix(zoom, zoom)

    for page in src:
        pix = page.get_pixmap(matrix=matrix)
        img_bytes = pix.tobytes("jpeg", jpg_quality=int(jpeg_quality))

        new_page = out.new_page(width=page.rect.width, height=page.rect.height)
        new_page.insert_image(new_page.rect, stream=img_bytes)

    result_bytes = out.tobytes(garbage=4, deflate=True)

    out.close()
    src.close()

    return result_bytes
```

```python
def convert_images_to_pdf(output_path, image_files):
    """
    Convert a list of image files into a single PDF, one image per page,
    optionally optimizing the result for smaller file size.

    :param output_path: Path to the output PDF. Input by the user.
    :param image_files: List of image file names, in the order they should
    appear in the PDF.
    Return a message.
    """
    if output_path in os.listdir():
        return "File " + output_path + " already exists. No action taken. Delete or rename it first."

    pdf_bytes = build_pdf_bytes(image_files)

    if optimize_pdf == "optimize":
        target_dpi = get_target_dpi()
        pdf_bytes = downsample_pdf_bytes(pdf_bytes, target_dpi, jpeg_quality)

    with open(output_path, 'wb') as output_file:
        output_file.write(pdf_bytes)

    image_files_str = ', \n'.join(image_files)

    return "Images converted:\n" + image_files_str + "\n\nSaved as:\n" + output_path + "\n\nRefresh the Files area and locate " + output_path + "."
```

### Run the conversion


```python
# Get the list of image files, sorted alphabetically
image_files, image_files_response = get_image_files()

if len(image_files) == 0:
    result = image_files_response
else:
    result = convert_images_to_pdf(output_pdf_path, image_files)
```
